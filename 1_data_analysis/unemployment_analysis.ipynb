# Unemployment Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_csv("data/unemployment.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
# change column names
df.columns = [ "State", "Date", "Frequency", "Estimated Unemployment Rate (%)", "Estimated Employed", "Estimated Labour Participation Rate (%)", "Region", "longitude", "latitude"]
```

```python
# let's see a corrolation between the variables

plt.style.use('seaborn-whitegrid')

plt.figure(figsize=(12, 10))

colums_to_consider = ["Estimated Unemployment Rate (%)", "Estimated Employed", "Estimated Labour Participation Rate (%)", "longitude", "latitude"]

correlation_matrix = df[colums_to_consider].corr()

sns.heatmap(correlation_matrix)
plt.title('Correlation Heatmap of Variables')
plt.show()
```

```python
# Estimated empoyed by region


plt.title("Estimated empoyed by region")
sns.histplot(x="Estimated Employed", hue="Region", data=df)
plt.show()
```

```python
unemployment =  df [ [ "State" ,  "Region" , "Estimated Unemployment Rate (%)"]]

fig = px.sunburst(unemployment , path = ["Region" , "State"], values = "Estimated Unemployment Rate (%)" , color_continuous_scale = "Viridis")

fig.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv("data/billionaire.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df["NetWorth"] = df["NetWorth"].str.replace("$" , "").str.replace("B" , "").astype(float)
```

```python
df_temp_sorted =  df.sort_values(by="NetWorth", ascending=False).head(10)

plt.figure(figsize=(10,6))
sns.histplot(data= df_temp_sorted , x = "Name" , hue = "NetWorth")
plt.show()
```

```python
top_domains = df["Source"].value_counts().head()

idx = top_domains.index
sources =  top_domains.values

custom_colours = ["#FFC300", "#88D8B0", "#FF5733", "#74C0FC", "#F95733"]
plt.figure(figsize=(12,8))
plt.pie(sources, labels = idx, colors = custom_colours,)

central_circle = plt.Circle((0,0),0.70,fc='white')

fig = plt.gcf()
fig.gca().add_artist(central_circle)

plt.show()
```

```python
top_5_industries = df["Industry"].value_counts().head()

idx =  top_5_industries.index
industries = top_5_industries.values

custom_colours = ["#FFC300", "#88D8B0", "#FF5733", "#74C0FC", "#F95733"]

plt.figure(figsize=(12,8))
plt.pie(industries, labels = idx, colors = custom_colours)
central_circle = plt.Circle((0,0),0.70,fc='white')
fig = plt.gcf()
fig.gca().add_artist(central_circle)
plt.rc('font', size=14)
plt.title("Top 5 Industries of billionaires", fontsize=20)

plt.show()
```

```python
a  =df["Country"].value_counts().head()
idx = a.index
countries =  a.values

custom_colours = ["skyblue", "yellowgreen", 'tomato', "blue", "red"]
plt.figure(figsize=(12,8))
plt.pie(countries, labels=idx, colors=custom_colours)
central_circle = plt.Circle((0,0),0.70,fc='white')
fig = plt.gcf()
fig.gca().add_artist(central_circle)
plt.show()
```
